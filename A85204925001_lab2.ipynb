## Exercise 10: Vectorized arithmetic, a small prediction model (10 marks)

Do **not** use `for` loops in this exercise.

A shop predicts its daily sales (in units) from two features: the temperature and the number of visitors. The model is $\hat{y} = X w + b$.

| Part | Task | Marks |
|---|---|---|
| (a) | `celsius_to_f`: convert the 1-D array `celsius` to Fahrenheit using $F = C \times 1.8 + 32$. | 2 |
| (b) | `squares`: the square of every value in `nums`, and `sum_squares`: their total. | 2 |
| (c) | `y_hat`: the predictions of the model for all 5 days, using `X`, `w` and `b`. | 2 |
| (d) | `errors`: the true values `y` minus the predictions `y_hat`. | 2 |
| (e) | `mse`: the Mean Squared Error, the average of the squared errors. | 2 |

````python
celsius = np.array([0, 10, 20, 30, 40])

nums = np.array([1, 2, 3, 4, 5, 6])

# 5 days, 2 features: temperature and visitors (in hundreds)
X = np.array([[20, 3],
              [25, 5],
              [18, 2],
              [30, 6],
              [22, 4]])
y = np.array([52, 75, 40, 96, 60])   # true sales

w = np.array([1.5, 8.0])
b = 2.0

celsius_to_f = celsius * 1.8 + 32  # YOUR CODE HERE

squares = np.square(nums)  # YOUR CODE HERE

sum_squares = squares.sum()  # YOUR CODE HERE

y_hat = X @ w + b  # YOUR CODE HERE

errors = y - y_hat  # YOUR CODE HERE

mse = np.mean(errors ** 2)  # YOUR CODE HERE

show("celsius_to_f", celsius_to_f)
show("squares", squares)
show("sum_squares", sum_squares)
show("y_hat", y_hat)
show("errors", errors)
show("mse", mse)
````
